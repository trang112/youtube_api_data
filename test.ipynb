```python
#lấy thông tin channel replyD, channel title và upload id
import requests
import pandas as pd
import json

handle = "@Sontungmtp"
base_url = "https://www.googleapis.com/youtube/v3"
api_key = "AIzaSyAoRcYaRdAyiFtompaNaEsXRl6fORD2zHg"

params = {
    "part":"id,snippet,contentDetails,statistics",
    "forHandle":handle,
    "key":api_key
}

response = requests.get(f"{base_url}/channels",params = params, timeout = 30)
response.raise_for_status()
data = response.json()
print(data)
channel_id = data["items"][0]["id"]
channel_title = data["items"][0]["snippet"]["title"]
video_count = data["items"][0]["statistics"]["videoCount"]
uploads_id = data["items"][0]["contentDetails"]["relatedPlaylists"]["uploads"]
print(channel_id)
print(video_count)
print(uploads_id)
```

```python
#get all information of video id through upload_id in the playlist items
video_id = []
page_token = None
while True:
    response2 = requests.get(f"{base_url}/playlistItems",params = {
        "part":"contentDetails",
        "playlistId": uploads_id,
        "key": api_key,
        "maxResults":50,
        "pageToken": page_token
    }, timeout = 30)
    response2.raise_for_status()
    data2 = response2.json()
    video_id += [i["contentDetails"]["videoId"] for i in data2["items"]]
    page_token = data2.get("nextPageToken")
    if not page_token:
        break
print(video_id)
```

```python
#get detailed information of video through video id
videos_raw = []
for start in range(0, len(video_id), 50):
    batch = video_id[start:start+50]
    r = requests.get(f"{base_url}/videos", timeout=30, params={
        "part": "snippet,statistics,contentDetails",
        "id": ",".join(batch),
        "key": api_key,
    })
    r.raise_for_status()
    data3 = r.json()
    videos_raw.append(data3)
print(len(videos_raw))
with open("data3.json", "w", encoding="utf-8") as f:
        json.dump(videos_raw, f, ensure_ascii=False, indent=2)
```

```python
#get all neccesary info in table video
result_video = []
for page in videos_raw: 
    for i in page["items"]:
        dict_video_info = {}
        dict_video_info["channel_id"] = i["snippet"]["channelId"]
        dict_video_info["channel_title"] = i["snippet"]["channelTitle"]
        dict_video_info["video_id"] = i["id"]
        dict_video_info["video_title"] = i["snippet"]["title"]
        dict_video_info["description"] = i["snippet"]["description"]
        dict_video_info["published_at"] = i["snippet"]["publishedAt"]
        dict_video_info["duration"] = i["contentDetails"]["duration"]
        dict_video_info["tags"] = i["snippet"].get("tags")
        dict_video_info["category_id"] = i["snippet"]["categoryId"]
        dict_video_info["view_count"] = i["statistics"]["viewCount"]
        dict_video_info["like_count"] = i["statistics"]["likeCount"]
        dict_video_info["comment_count"] = i["statistics"]["commentCount"]
        result_video.append(dict_video_info) 

df = pd.DataFrame(result_video)
print(df)
```

```python
#get video_id of top 10 videos with highest view count 
df["view_count"] = pd.to_numeric(df["view_count"], errors="coerce")
top_10 = df.nlargest(10,"view_count")["video_id"].tolist()
print(top_10)
```

```python
#getting data from comment thread
comment = []
for vid in top_10:
    page_token2 = None
    page = 0
    while page < 20:
        r_comment = requests.get(f"{base_url}/commentThreads", timeout=30, params={
            "part": "snippet,replies",
            "videoId": vid,
            "maxResults": 100,
            "order": "time",
            "textFormat": "plainText",
            "pageToken":page_token2,
            "key": api_key,
        })
        r_comment.raise_for_status()
        data_comment = r_comment.json()
        for i in data_comment["items"]:
            dict_c_info = {}
            dict_c_info["channel_id"] = i["snippet"]["channelId"]
            dict_c_info["video_id"] = i["snippet"]["videoId"]
            dict_c_info["comment_id"] = i["snippet"]["topLevelComment"]["id"]
            dict_c_info["parent_id"] = None
            dict_c_info["author_name"] = i["snippet"]["topLevelComment"]["snippet"]["authorDisplayName"]
            dict_c_info["comment_text"] = i["snippet"]["topLevelComment"]["snippet"]["textDisplay"]
            dict_c_info["published_at"] = i["snippet"]["topLevelComment"]["snippet"]["publishedAt"]
            dict_c_info["updated_at"] = i["snippet"]["topLevelComment"]["snippet"]["updatedAt"]
            dict_c_info["like_count"] = i["snippet"]["topLevelComment"]["snippet"]["likeCount"]
            dict_c_info["reply_count"] = i["snippet"]["totalReplyCount"]
            comment.append(dict_c_info)

            for r in i.get("replies",{}).get("comments",[]):
                dict_r_info = {}
                dict_r_info["channel_id"] = r["snippet"]["channelId"]
                dict_r_info["video_id"] = r["snippet"]["videoId"]
                dict_r_info["comment_id"] = r["id"]
                dict_r_info["parent_id"] = r["snippet"]["parentId"]
                dict_r_info["author_name"] = r["snippet"]["authorDisplayName"]
                dict_r_info["comment_text"] = r["snippet"]["textDisplay"]
                dict_r_info["published_at"] = r["snippet"]["publishedAt"]
                dict_r_info["updated_at"] = r["snippet"]["updatedAt"]
                dict_r_info["like_count"] = r["snippet"]["likeCount"]
                dict_r_info["reply_count"] = None
                comment.append(dict_r_info)
        page += 1
        page_token2 = data_comment.get("nextPageToken")
        
        if not page_token2:
            break
print(data_comment)
with open("data_comment.json", "w", encoding="utf-8") as f:
        json.dump(comment, f, ensure_ascii=False, indent=2)

df1 = pd.DataFrame(comment)
df1.to_csv("data_comment.csv",index=False, encoding='utf-8-sig')
print(df1)
```
